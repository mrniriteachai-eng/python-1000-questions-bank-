## 🟠 EXCEPTION HANDLING QUESTIONS

2️⃣1️⃣ Basic Try Except

```python
try:
    print(10 / 0)

except:
    print("Error")
```

2️⃣2️⃣ Handle Value Error

```python
try:
    num = int(input())

except ValueError:
    print("Invalid Input")
```

2️⃣3️⃣ Handle ZeroDivisionError

```python
try:
    print(10 / 0)

except ZeroDivisionError:
    print("Cannot Divide by Zero")
```

2️⃣4️⃣ Handle File Error

```python
try:

    file = open("abc.txt", "r")

except FileNotFoundError:
    print("File Not Found")
```

2️⃣5️⃣ Else Block

```python
try:
    print(10 / 2)

except:
    print("Error")

else:
    print("Success")
```

2️⃣6️⃣ Finally Block

```python
try:
    print(10 / 2)

except:
    print("Error")

finally:
    print("Always Runs")
```

2️⃣7️⃣ Multiple Exceptions

```python
try:

    num = int(input())

    print(10 / num)

except ValueError:
    print("Invalid Number")

except ZeroDivisionError:
    print("Cannot Divide by Zero")
```

2️⃣8️⃣ Exception as Variable

```python
try:
    print(10 / 0)

except Exception as e:
    print(e)
```

2️⃣9️⃣ Custom Error Message

```python
try:

    age = int(input())

    if age < 0:
        raise ValueError

except ValueError:
    print("Invalid Age")
```

## 🔴 ADVANCED QUESTIONS

4️⃣1️⃣ File Word Frequency

```python
with open("demo.txt", "r") as file:

    text = file.read()

words = text.split()

freq = {}

for word in words:

    if word in freq:
        freq[word] += 1

    else:
        freq[word] = 1

print(freq)
```

4️⃣2️⃣ Largest Line

```python
with open("demo.txt", "r") as file:

    lines = file.readlines()

largest = max(lines, key=len)

print(largest)
```

4️⃣3️⃣ Smallest Line

```python
with open("demo.txt", "r") as file:

    lines = file.readlines()

smallest = min(lines, key=len)

print(smallest)
```

4️⃣4️⃣ Encrypt Text

```python
text = "python"

encrypted = ""

for char in text:
    encrypted += chr(ord(char) + 1)

print(encrypted)
```

4️⃣5️⃣ Decrypt Text

```python
text = "qzuipo"

decrypted = ""

for char in text:
    decrypted += chr(ord(char) - 1)

print(decrypted)
```
