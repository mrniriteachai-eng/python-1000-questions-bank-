🟢 BASIC LIST QUESTIONS

1️⃣ Create List

```python
num = [1,2,3,4]
print(num)
```

2️⃣ Access First Element

```python
num = [1,2,3,4]
print(num[0])
```

3️⃣ Access Last Element

```python
num = [1,2,3,4]
print(num[-1])
```

4️⃣ List Length

```python
nums = [1, 2, 3, 4, 5]
print(len(nums))
```

5️⃣ Change List Value

```python
num = [1,2,3,4,5]
num[0] = 100
print(num)
```

6️⃣ Append Item

```python
num = [1,2,3,4]
num.append(6)
print(num)
```

7️⃣ Insert Item

```python
num = [1,2,3,4]
num.insert(1,3)
print(num)
```

8️⃣ Remove Item

```python
nums = [1,2,3,4]
nums.remove(2)
print(nums)
```

9️⃣ Pop Item

```python
nums = [1, 2, 3]

nums.pop()

print(nums)
```

🔟 Clear List

```python
nums = [1,2,3,4]
nums.clear()
print(nums)
```
