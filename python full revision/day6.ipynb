## 🔵 ADVANCED LIST QUESTIONS

3️⃣1️⃣ Largest Number Manual

```python
nums = [10, 50, 20]

largest = nums[0]

for i in nums:

    if i > largest:
        largest = i

print(largest)
```

3️⃣2️⃣ Smallest Number Manual

```python
nums = [10, 50, 20]

smallest = nums[0]

for i in nums:

    if i < smallest:
        smallest = i

print(smallest)
```

3️⃣3️⃣ Sum Without sum()

```python
nums = [1, 2, 3]

total = 0

for i in nums:
    total += i

print(total)
```

3️⃣4️⃣ Average of List

```python
nums = [10, 20, 30]

avg = sum(nums) / len(nums)

print(avg)
```

3️⃣5️⃣ Second Largestv

```python
nums = [10, 50, 20, 40]

nums.sort()

print(nums[-2])
```

3️⃣6️⃣ Second Smallest

```python
nums = [50, 60, 70, 80]
nums.sort()
print(nums[1])
```

3️⃣7️⃣ Frequency of Elements

```python
nums = [1, 2, 2, 3]

for i in set(nums):
    print(i, nums.count(i))
```

3️⃣8️⃣ Common Elements

```python
a = [1, 2, 3]
b = [2, 3, 4]

for i in a:

    if i in b:
        print(i)
```

3️⃣9️⃣ Difference Between Lists

```python
a = [1, 2, 3]
b = [2, 3]

for i in a:

    if i not in b:
        print(i)
```

4️⃣0️⃣ List Comprehension

```python
nums = [1, 2, 3, 4]

square = [i ** 2 for i in nums]

print(square)
```

## 🔴 NESTED LOOP LIST QUESTIONS

4️⃣1️⃣ Matrix Print

```python
matrix = [
    [1, 2],
    [3, 4]
]

for row in matrix:
    print(row)
```

4️⃣2️⃣ Matrix Elements

```python
matrix = [
    [1, 2],
    [3, 4]
]

for row in matrix:

    for item in row:
        print(item)
```

4️⃣3️⃣ Sum Matrix

```python
matrix = [
    [1, 2],
    [3, 4]
]

total = 0

for row in matrix:

    for item in row:
        total += item

print(total)
```

4️⃣4️⃣ Flatten List

```python
matrix = [
    [1, 2],
    [3, 4]
]

flat = []

for row in matrix:

    for item in row:
        flat.append(item)

print(flat)
```

4️⃣5️⃣ Multiplication Table List

```python
nums = [1, 2, 3, 2, 1]

if nums == nums[::-1]:
    print("Palindrome")

else:
    print("Not")
```

4️⃣7️⃣ Rotate List

```python
nums = [1, 2, 3, 4]

rotated = nums[1:] + nums[:1]

print(rotated)
```

4️⃣8️⃣ Separate Even Odd

```python
nums = [1, 2, 3, 4]

even = []
odd = []

for i in nums:

    if i % 2 == 0:
        even.append(i)

    else:
        odd.append(i)

print(even)
print(odd)
```

4️⃣9️⃣ Remove Negative Numbers

```python
nums = [1, -2, 3, -4]

new = []

for i in nums:

    if i > 0:
        new.append(i)

print(new)
```
