2️⃣2️⃣ Sum 1 to N

```python
def total(n):

    sum = 0

    for i in range(1, n + 1):
        sum += i

    return sum

print(total(10))
```

2️⃣3️⃣ Factorial Function

```python
def factorial(num):

    fact = 1

    for i in range(1, num + 1):
        fact *= i

    return fact

print(factorial(5))
```

2️⃣4️⃣ Prime Function

```python
def prime(num):

    for i in range(2, num):

        if num % i == 0:
            return "Not Prime"

    return "Prime"

print(prime(7))
```

2️⃣5️⃣ Fibonacci Function

```python
def fibonacci():

    a = 0
    b = 1

    for i in range(10):
        print(a)

        a, b = b, a + b

fibonacci()
```

2️⃣6️⃣ Count Vowels

```python
def vowels(text):

    count = 0

    for char in text:

        if char in "aeiou":
            count += 1

    return count

print(vowels("python"))
```

2️⃣7️⃣ Palindrome Function

```python
def palindrome(text):

    if text == text[::-1]:
        return "Palindrome"

    return "Not"

print(palindrome("madam"))
```

## 🔥 ADVANCED FUNCTION QUESTIONS

4️⃣1️⃣ Calculator Function

```python
def calculator(a, b, op):

    if op == "+":
        return a + b

    elif op == "-":
        return a - b

    elif op == "*":
        return a * b

    elif op == "/":
        return a / b

print(calculator(10, 5, "+"))
```

4️⃣2️⃣ BMI Function

```python
def bmi(weight, height):

    return weight / (height ** 2)

print(bmi(60, 1.7))
```

4️⃣3️⃣ Percentage Function

```python
def percentage(obtained, total):

    return (obtained / total) * 100

print(percentage(450, 500))
```

4️⃣4️⃣ Login Function

```python
def login(username, password):

    if username == "admin" and password == "1234":
        return "Login Success"

    return "Login Failed"

print(login("admin", "1234"))
```

4️⃣5️⃣ Email Validation

## 🚀 EXPERT BEGINNER QUESTIONS

4️⃣6️⃣ Recursive Function

```python
def count(n):

    if n == 0:
        return

    print(n)

    count(n - 1)

count(5)
```

4️⃣7️⃣ Recursive Factorial

```python
def factorial(n):

    if n == 1:
        return 1

    return n * factorial(n - 1)

print(factorial(5))
```

4️⃣8️⃣ Recursive Sum

```python
def total(n):

    if n == 0:
        return 0

    return n + total(n - 1)

print(total(5))
```

4️⃣9️⃣ Recursive Fibonacci

```python
def fib(n):

    if n <= 1:
        return n

    return fib(n-1) + fib(n-2)

print(fib(6))
```
