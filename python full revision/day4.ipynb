🟡 INTERMEDIATE LOOP QUESTIONS

1️⃣1️⃣ Count Digits

```python
num = 12345

count = 0

while num > 0:
    count += 1
    num //= 10

print(count)
```

1️⃣2️⃣ Reverse Number

```python
num = 1234

rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num //= 10

print(rev)
```

1️⃣3️⃣ Palindrome Number

```python
num = 121
temp = num
rev = 0

while num > 0:
    digit = num % 10
    rev = rev * 10 + digit
    num //= 10

if temp == rev:
    print("Palindrome")

else:
    print("Not")
```

## 🔴 ADVANCED LOOP QUESTIONS

3️⃣6️⃣ Armstrong Number

```python
num = 153
temp = num
total = 0

while num > 0:
    digit = num % 10
    total += digit ** 3
    num //= 10

if temp == total:
    print("Armstrong")

else:
    print("Not Armstrong")
```

3️⃣7️⃣ Strong Number

```python
num = 145
temp = num
total = 0

while num > 0:

    digit = num % 10
    fact = 1

    for i in range(1, digit + 1):
        fact *= i

    total += fact
    num //= 10

if temp == total:
    print("Strong Number")

else:
    print("Not")
```

3️⃣8️⃣ Prime Numbers 1–100

```python
for num in range(2, 101):

    prime = True

    for i in range(2, num):

        if num % i == 0:
            prime = False

    if prime:
        print(num)
```

## 🔥 PATTERN QUESTIONS

4️⃣1️⃣ Pyramid Pattern

```python
for i in range(1, 6):

    print(" " * (5-i) + "*" * (2*i-1))
```

4️⃣2️⃣ Inverted Pyramid

```python
for i in range(5, 0, -1):

    print(" " * (5-i) + "*" * (2*i-1))
```

4️⃣3️⃣ Hollow Square

```python
for i in range(5):

    for j in range(5):

        if i == 0 or i == 4 or j == 0 or j == 4:
            print("*", end=" ")

        else:
            print(" ", end=" ")

    print()
```

4️⃣4️⃣ Floyd Triangle

```python
num = 1

for i in range(1, 6):

    for j in range(i):
        print(num, end=" ")
        num += 1

    print()
```

4️⃣5️⃣ Alphabet Pattern

```python
for i in range(65, 70):

    for j in range(65, i+1):
        print(chr(j), end=" ")

    print()
```
