## 🔴 ADVANCED STRING QUESTIONS

4️⃣1️⃣ Frequency of Characters

```python
text = "banana"

for char in set(text):
    print(char, text.count(char))
```

4️⃣2️⃣ Most Repeated Character

```python
text = "banana"

max_char = ""

max_count = 0

for char in text:

    if text.count(char) > max_count:
        max_count = text.count(char)
        max_char = char

print(max_char)
```

4️⃣3️⃣ Remove Duplicate Characters

```python
text = "programming"

new = ""

for char in text:

    if char not in new:
        new += char

print(new)
```

4️⃣4️⃣ Anagram Check

```python
a = "listen"
b = "silent"

if sorted(a) == sorted(b):
    print("Anagram")

else:
    print("Not")
```

4️⃣5️⃣ String Compression

```python
text = "aaabb"

count = 1

for i in range(len(text)-1):

    if text[i] == text[i+1]:
        count += 1

    else:
        print(text[i] + str(count))
        count = 1

print(text[-1] + str(count))
```

4️⃣9️⃣ Count Upper Lower

```python
text = "PyThOn"

upper = 0
lower = 0

for char in text:

    if char.isupper():
        upper += 1

    elif char.islower():
        lower += 1

print("Upper:", upper)
print("Lower:", lower)
```
