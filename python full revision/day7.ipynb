## 🟢 TUPLE QUESTIONS


1️⃣ Create Tuple

```python
data = (1, 2, 3)

print(data)
```

2️⃣ Access First Item

```python
data = (10, 20, 30)

print(data[0])
```

3️⃣ Access Last Item

```python
data = (10,20,30)
print(data[-1])
```

4️⃣ Tuple Length

```python
data = (10,20,30)
print(len(data))
```

5️⃣ Loop Tuple

```python
data = (1, 2, 3)

for i in data:
    print(i)
```

7️⃣ Find Index

```python
nums = {1, 2, 3}

print(nums)
```

## 🟠 DICTIONARY QUESTIONS

2️⃣1️⃣ Create Dictionary

```python
student = {
    "name" : "nirmal",
    "age" : 18
}

print(student)
```

2️⃣2️⃣ Access Value

```python
student = {
    "name" : "nirmal",
    "age" : 18
}

print(student["name"])
```

2️⃣3️⃣ Change Value

```python
student = {
    "age": 18
}

student["age"] = 20

print(student)
```

2️⃣4️⃣ Add New Key

```python
student = {
    "name": "Nirmal"
}

student["age"] = 18

print(student)
```

2️⃣5️⃣ Remove Key

```python
student = {
    "name": "Nirmal",
    "age": 18
}

student.pop("age")

print(student)
```

2️⃣6️⃣ Loop Dictionary Keys

```python
student = {
    "name": "Nirmal",
    "age": 18
}

for key in student:
    print(key)
```

2️⃣7️⃣ Loop Dictionary Values

```python
student = {
    "name": "Nirmal",
    "age": 18
}

for value in student.values():
    print(value)
```

2️⃣8️⃣ Loop Key and Value

```python
student = {
    "name": "Nirmal",
    "age": 18
}

for key, value in student.items():
    print(key, value)
```

2️⃣9️⃣ Dictionary Length

```python
student = {
    "name": "Nirmal",
    "age": 18
}

print(len(student))
```

3️⃣0️⃣ Check Key Exists

```python
student = {
    "name": "Nirmal"
}

print("name" in student)
```

## 🔵 INTERMEDIATE QUESTIONS

3️⃣1️⃣ Nested Dictionary

```python
student = {
    "info": {
        "name": "Nirmal",
        "age": 18
    }
}

print(student["info"]["name"])
```

3️⃣2️⃣ Dictionary from User Input

```python
name = input("Enter name: ")
age = int(input("Enter age: "))

student = {
    "name": name,
    "age": age
}

print(student)
```

3️⃣3️⃣ Frequency Counter

```python
text = "banana"

freq = {}

for char in text:

    if char in freq:
        freq[char] += 1

    else:
        freq[char] = 1

print(freq)
```

3️⃣4️⃣ Square Dictionary

```python
nums = {}

for i in range(1, 6):
    nums[i] = i ** 2

print(nums)
```

3️⃣5️⃣ Merge Dictionaries

```python
a = {"x": 1}
b = {"y": 2}

a.update(b)

print(a)
```

3️⃣6️⃣ Largest Value in Dictionary

```python
marks = {
    "math": 90,
    "science": 80
}

print(max(marks.values()))
```

3️⃣7️⃣ Smallest Value

```python
marks = {
    "math": 90,
    "science": 80
}

print(min(marks.values()))
```

3️⃣9️⃣ Sort Dictionary Keys

```python
student = {
    "b": 2,
    "a": 1
}

for key in sorted(student):
    print(key)
```

4️⃣0️⃣ Dictionary Comprehension

```python


square = {x: x**2 for x in range(1, 6)}

print(square)
```

## 🔴 ADVANCED QUESTIONS

4️⃣1️⃣ Count Vowels Using Dictionary

```python
text = "python programming"

count = {
    "a": 0,
    "e": 0,
    "i": 0,
    "o": 0,
    "u": 0
}

for char in text:

    if char in count:
        count[char] += 1

print(count)
```

4️⃣2️⃣ Common Elements Using Set

```python
a = {1, 2, 3}
b = {2, 3, 4}

print(a & b)
```

4️⃣3️⃣ Remove Duplicate Words

```python
words = ["python", "java", "python"]

print(list(set(words)))
```

4️⃣4️⃣ Student Marks System

```python
marks = {
    "math": 80,
    "science": 90,
    "english": 70
}

total = sum(marks.values())

print(total)
```

4️⃣5️⃣ Highest Marks Subject

```python
marks = {
    "math": 80,
    "science": 95,
    "english": 70
}

print(max(marks, key=marks.get))
```
